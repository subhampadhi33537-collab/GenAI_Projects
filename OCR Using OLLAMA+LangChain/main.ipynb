```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_core.messages import HumanMessage,AIMessage
```

```python
model=ChatOllama(
    model="ollamalens",
    temperature=0.2
)
```

```python
image1_path="images/Screenshot (73).png"
image2_path="images/WhatsApp Image 2026-09-28 at 19.50.28.jpeg"
image3_path="images/WhatsApp Image 2026-09-28 at 19.51.15.jpeg"
```

```python
import base64

with open(image1_path, "rb") as f:
    image1_base64 = base64.b64encode(f.read()).decode("utf-8")

with open(image2_path, "rb") as f:
    image2_base64 = base64.b64encode(f.read()).decode("utf-8")

with open(image3_path, "rb") as f:
    image3_base64 = base64.b64encode(f.read()).decode("utf-8")
```

```python
message1 = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Extract all text from this image using OCR."
        },
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{image1_base64}"
            }
        }
    ]
)

message2 = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Extract all text from this image using OCR."
        },
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{image2_base64}"
            }
        }
    ]
)

message3 = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Extract all text from this image using OCR."
        },
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/png;base64,{image3_base64}"
            }
        }
    ]
)
```

```python
parser=StrOutputParser()

chain=model | parser
```

```python
chain.invoke(message1)
```
