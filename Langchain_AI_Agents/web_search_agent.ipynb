```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
import requests
```

```python
search_tool=DuckDuckGoSearchRun()

search_tool.invoke("what is the name  of the education minister of india")
```

```python
llm=ChatGroq(model="openai/gpt-oss-120b")
```

```python
llm.invoke("hii")
```

```python
from langchain.agents import create_agent
```

```python
system_prompt = """
You are a helpful web search assistant.

Your job is to answer the user's questions using the DuckDuckGo search tool when
the question requires up-to-date, factual, or web-based information.

Instructions:
1. Use the DuckDuckGo search tool to search for information relevant to the user's query.
2. Do not use the search tool for simple questions that you can answer confidently without web search.
3. When searching, use a clear and concise search query based on the user's request.
4. Carefully analyze the search results before answering.
5. Provide a concise and accurate answer based on the search results.
6. If the search results do not contain enough information, clearly say that the information could not be found.
7. Do not invent or hallucinate information that is not supported by the search results.
"""
```

```python
agent=create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt=system_prompt
)
```

```python
result = agent.invoke({"messages": [{"role": "user", 
                                     "content": "what is langsmith?"}]})
```
